```python
df_funds['Funds Sanctioned (Rs. In Cr.)'] = pd.to_numeric(df_funds['Funds Sanctioned (Rs. In Cr.)'])
df_funds['Funds Utilized (Rs. In Cr.)'] = pd.to_numeric(df_funds['Funds Utilized (Rs. In Cr.)'])
```

```python
columns_to_convert = [col for col in df_state_data.columns if 'Opened' in col or 'Closed' in col]

for col in columns_to_convert:
    df_state_data[col] = pd.to_numeric(df_state_data[col])

# print("df_state_data info after type conversion:")
# df_state_data.info()
# print("\ndf_state_data head after type conversion:")
# print(df_state_data.head())
```

```python
df_funds_melted = df_funds.melt(id_vars=['Sl. No.', 'Financial Year'], value_vars=['Funds Sanctioned (Rs. In Cr.)', 'Funds Utilized (Rs. In Cr.)'], var_name='Metric', value_name='Value')

custom_palette = ['#D7C4F3', '#F6B8D8']

plt.figure(figsize=(8, 5))

ax = sns.barplot(x='Financial Year', y='Value', hue='Metric', data=df_funds_melted, palette=custom_palette, width=0.4)

# Show exact values on each bar
for bar in ax.patches:
    height = bar.get_height()
    ax.annotate(f'{height:.1f}', (bar.get_x() + bar.get_width() / 2, height), ha='center', va='bottom', fontsize=10)

plt.title('Funds Sanctioned vs. Funds Utilized by Financial Year', fontsize=14)
plt.xlabel('Financial Year')
plt.ylabel('Amount (Rs. In Cr.)')

plt.legend(title='Funds Type', fontsize=8)

sns.despine()
plt.tight_layout()
plt.grid(axis='y', alpha=0.2)
plt.show()
```

```python
df_state_data_melted = df_state_data.melt(id_vars=['Sl. No.', 'State/UT'], value_vars=columns_to_convert, var_name='Metric', value_name='Count')

df_state_data_melted['Year'] = df_state_data_melted['Metric'].str.extract(r'(\d{4}-\d{4})')
df_state_data_melted['Type'] = df_state_data_melted['Metric'].str.split(' - ').str[0]

# print("df_state_data_melted head:")
# print(df_state_data_melted.head())
# print("\ndf_state_data_melted info:")
# df_state_data_melted.info()
```

```python
df_yearly_summary = df_state_data_melted.groupby(['Year', 'Type'])['Count'].sum().reset_index()

print("df_yearly_summary head:")
print(df_yearly_summary.head())
```

```python
custom_palette = ['#D7C4F3', '#F6B8D8']

plt.figure(figsize=(9, 6))

ax = sns.barplot(x='Year', y='Count', hue='Type', data=df_yearly_summary, palette=custom_palette, width=0.45)

for bar in ax.patches:
    height = bar.get_height()
    ax.annotate(str(int(height)), (bar.get_x() + bar.get_width() / 2, height), ha='center', va='bottom', fontsize=10)

plt.title('Total Opened and Closed PMBJKs by Year', fontsize=14)
plt.xlabel('Year')
plt.ylabel('Number of PMBJKs')

plt.legend(title='Type')

sns.despine()
plt.grid(axis='y', alpha=0.2)
plt.tight_layout()
plt.show()
```
